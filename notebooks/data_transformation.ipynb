<p style='font-size: 1.4em;'><u><b style='color: red;'>Step 2.1.</b><b> Add Columns That Is Related to Time</b></u></p>

```python
# Force Pandas to recognize the 'Sale Order Date' column as a datetime object for pandas to do further analysis.
df['Sales Order Date'] = pd.to_datetime(df['Sales Order Date'])
```

<p style='font-size: 1.4em;'><u><b style='color: orange;'>Step 2.1.1.</b><b> Adding 'Year' and 'Month'</b></u></p>

```python
# Add column 'Year' and 'Month'.
# 1) Adding the 'Year' column.
df['Year'] = df['Sales Order Date'].dt.year

# 2) Adding the 'Month' column.
df['Month'] = df['Sales Order Date'].dt.month
df
```

<p style='font-size: 1.4em;'><u><b style='color: orange;'>Step 2.1.2.</b><b> Adding 'Quarter'</b></u></p>

```python
# Adding the 'Quarter' column.
df['Quarter'] = df['Sales Order Date'].dt.quarter
df
```

<p style='font-size: 1.4em;'><u><b style='color: orange;'>Step 2.1.3.</b><b> Adding 'is_month_end' and 'is_quarter_end'</b></u></p>

```python
# Adding column 'is_month_end' and 'is_quarter_end'.
# 1) Adding the 'is_month_end' column.
df['is_month_end'] = df['Sales Order Date'].dt.is_month_end

# 2) Adding the 'is_quarter_end' column.
df['is_quarter_end'] = df['Sales Order Date'].dt.is_quarter_end
df
```

<p style='font-size: 1.4em;'><u><b style='color: orange;'>Step 2.1.4.</b><b> Adding 'Day of Week'</b></u></p>

```python
# Adding the 'Day of Week' column.
df['Day of Week'] = df['Sales Order Date'].dt.day_of_week
df
```

<p style='font-size: 1.4em;'><u><b style='color: orange;'>Step 2.1.5.</b><b> Adding 'is_weekend'</b></u></p>

```python
# Adding the 'is_weekend' column.
df['is_weekend'] = df['Sales Order Date'].dt.day_of_week >= 5
df
```

<p style='font-size: 1.4em;'><u><b style='color: orange;'>Step 2.1.6.</b><b> Adding 'Day of Month'</b></u></p>

```python
# Adding the 'Day of Month' column.
df['Day of Month'] = df['Sales Order Date'].dt.day
df
```

<p style='font-size: 1.4em;'><u><b style='color: orange;'>Step 2.1.7.</b><b> Adding 'Year-Month'</b></u></p>

```python
# Adding 'Year-Month' column.
df['Year-Month'] = df['Sales Order Date'].dt.to_period('M')
df
```

---

<p style='font-size: 1.4em;'><u><b style='color: red;'>Step 2.2.</b><b> Add Columns That Is Related to Financial</b></u></p>

<p style='font-size: 1.4em;'><u><b style='color: orange;'>Step 2.2.1.</b><b> Add 'Total Sales Amount'</b></u></p>

```python
# Adding 'Total Sales Amount' column.
df['Total Sales Amount'] = df['Quantity'] * df['Base Price']
df
```

<p style='font-size: 1.4em;'><u><b style='color: orange;'>Step 2.2.2.</b><b> Add 'Discount Amount'</b></u></p>

```python
# Adding 'Discount Amount' column, don't multiply with a 100 yet.
df['Discount Amount'] = df['List Price'] - df['Base Price']
df
```

<p style='font-size: 1.4em;'><u><b style='color: orange;'>Step 2.2.3.</b><b> Add 'Discount Percentage'</b></u></p>

```python
# Adding 'Discount Percentage' column.
df['Discount Percentage'] = (df['Discount Amount'] / df['List Price'])
df
```

<p style='font-size: 1.4em;'><u><b style='color: orange;'>Step 2.2.4.</b><b> Add 'Total Gross Value' (Expected Revenue Before Discount)</b></u></p>

```python
# Adding 'Total Gross Value' column.
df['Total Gross Value'] = df['Quantity'] * df['List Price']
df
```

<p style='font-size: 1.4em;'><u><b style='color: orange;'>Step 2.2.5.</b><b> Add 'Total Discount Value' (Row-Level Revenue Loss)</b></u></p>

```python
# Adding 'Total Discount Value' column.
df['Total Discount Value'] = df['Quantity'] * df['Discount Amount']
df
```

<p style='font-size: 1.4em;'><u><b style='color: orange;'>Step 2.2.6.</b><b> Unit Selling Price (Effective Price)</b></u></p>

```python
# Adding 'Unit Selling Price' column.
df['Unit Selling Price'] = df['Total Sales Amount'] / df['Quantity']
df
```

<p style='font-size: 1.4em;'><u><b style='color: orange;'>Step 2.2.7.</b><b> Tiered Order Value Category (Order Size Bins)</b></u></p>

```python
# Define thresholds from bin labels.
bins = [-np.inf, 1000, 10_000, 50_000, np.inf]          # in Baht.
labels = ['Small (<1k)', 'Medium (1k-10k)', 'Large (10k-50k)', 'Enterprise (>50k)']

# Create the categorical order size feature
df['Order Size Tier'] = pd.cut(df['Total Sales Amount'], bins=bins, labels=labels)
df
```

<p style='font-size: 1.4em;'><u><b style='color: orange;'>Step 2.2.8.</b><b> Historical Revenue Features (Customer LTV (Customer Lifetime Value) & Average Order Value)</b></u></p>

<p style='font-size: 1.4em; l'><b>A. Cumulative Revenue (Customer Lifetime Value to Date)</b></p>
<p style='font-size: 1.4em; l'>Calculates the total revenue a customer has generated up to (and including) each transaction:</p>

```python
# Sort by customer and date to maintain chronological order.
df = df.sort_values(by=['Customer Internal Id', 'Sales Order Date'])

# Calculate cumulative historical revenue per customer.
df['Customer Cumulative Revenue'] = df.groupby('Customer Internal Id')['Total Sales Amount'].cumsum()
df
```

<p style='font-size: 1.4em; l'><b>B. Customer Average Order Value (AOV)</b></p>
<p style='font-size: 1.4em; l'>Calculates the historical average spend per transaction for each customer up to that point in time:</p>

```python
# Calculate cumulative order count per customer.
df['Customer Order Count'] = df.groupby('Customer Internal Id').cumcount() + 1

# Calculate expanding Average Order Value (AOV).
df['Customer AOV'] = df['Customer Cumulative Revenue'] / df['Customer Order Count']
df
```

<p style='font-size: 1.4em; l'><b>C. Overall Customer Lifetime Value (Static Snapshot)</b></p>
<p style='font-size: 1.4em; l'>To determine a single total summary metric per customer across the whole dataset (for segmentation rather than time-series forecasting)</p>

```python
# Total LTV snapshot.
customer_ltv = df.groupby('Customer Internal Id')['Total Sales Amount'].sum().rename('Customer Total LTV')
df = df.merge(customer_ltv, on='Customer Internal Id', how='left')
df
```

<p style='font-size: 1.4em;'><u><b style='color: orange;'>Step 2.2.9.</b><b> Pricing Variance / Discount Flag</b></u></p>

<p style='font-size: 1.4em; l'><b>1. Is_Discounted</b></p>
<p style='font-size: 1.4em; l'>Binary flag: 1 = Discounted Amount > 0, 0 = Else.</p>

```python
# Add 'is_discounted' column.
df['is_discounted'] = df['Discount Amount'] > 0
df
```

<p style='font-size: 1.4em; l'><b>2. Is_Zero_Base_Price</b></p>
<p style='font-size: 1.4em; l'>Binary flag: 1 = Base Price is 0, 0 = Else. Representing promotional bundles/ free samples/ negotiating price.</p>

```python
# Add 'is_zero_base_price' column.
df['is_zero_base_price'] = df['Base Price'] == 0
df
```

---

<p style='font-size: 1.4em;'><u><b style='color: red;'>Step 2.3.</b><b> Add Column of Refined the 'Sale Order Status' Column</b></u></p>

<p style='font-size: 1.4em;'>To simplify these 6 status for business logic or to have a simplified values. We can group these 6 values into the following table:</p>

| Simplified | Included System Status | Financial Meaning |
| :--- | :--- | :--- |
| Recognized / Billed Revenue | Billed | Invoiced and official income |
| Unbilled / Earned Pipeline | Pending Billing, Pending Billing/Partially Fulfilled | Work/goods delivered, pending invoice. Safe to count as projected income |
| Unearned / Open Backlog | Pending Fulfillment, Partially Fulfilled, Closed | Future potential sales, Unfulfilled items, or canceled orders. Do not count as income |

```python
# Define matching condition.
earned = df['Sales Order Status'].isin(['Billed'])
on_going = df['Sales Order Status'].isin(['Pending Billing', 'Pending Billing/Partially Fulfilled'])
loss = df['Sales Order Status'].isin([
    'Pending Fulfillment',
    'Partially Fulfilled',
    'Closed'
])

conditions = [earned, on_going, loss]
choices = ['Earned', 'On going', 'Loss']

# Apply Mapping.
df['Sales Order Status Group'] = np.select(conditions, choices, default='Other')
df
```

---

<p style='font-size: 1.4em;'><u><b style='color: red;'>Step 2.4.</b><b> Adding 'Day Since Last Purchase Column'</b></u></p>

```python
df = df.sort_values(['Customer Internal Id', 'Sales Order Date'])

# Days elapsed since previous order for each customer.
df['Days Since Last Purchase'] = (
    df.groupby('Customer Internal Id')['Sales Order Date'].diff().dt.days.fillna(0)
)
df
```

---

<p style='font-size: 1.4em;'><u><b style='color: red;'>Step 2.5.</b><b> Adding 'Product-Level Demand'</b></u></p>

<p style='font-size: 1.4em; l'><b>1. Item Historical Average Demand (Mean quantity sold per order for each SKU).</b></p>

```python
# Mean quantity sold per order for each SKU.
df['Product-Level Demand'] = df.groupby('Item Code')['Quantity'].transform('mean')
df
```

<p style='font-size: 1.4em; l'><b>2. Item Price Elasticity Ratio (Actual Base Price relative to the item's global mean price).</b></p>

```python
item_mean_price = df.groupby('Item Code')['Base Price'].transform('mean')
df['Item Price Elasticity Ratio'] = np.where(item_mean_price > 0, df['Base Price'] / item_mean_price, 1.0)
df
```

---

<p style='font-size: 1.4em;'><u><b style='color: red;'>Step 2.6.</b><b> Customer Product Diversity (Behavioral Signal)</b></u></p>

<p style='font-size: 1.4em; l'><b>1. Customer Unique Item Count (Cumulative count of distinct items bought up to that order date).</b></p>

```python
df['Customer Unique Item Count'] = (
    df.groupby('Customer Internal Id')['Item Code'].transform(lambda x: (~x.duplicated()).cumsum())
)
df
```

<p style='font-size: 1.4em; l'><b>2. Total unique items ever purchased by the customer across all history.</b></p>

```python
customer_total_unique = df.groupby('Customer Internal Id')['Item Code'].transform('nunique')
df['Customer Total Unique Items Overall'] = customer_total_unique
df
```

<p style='font-size: 1.4em;'><b style='color: Green;'><u>NOTE!</u></b></p>
<ol style='font-size: 1.4em; line-height: 1.4;'>
    <li><mark><b>Item Historical Average Demand</b></mark>: Gives tree-based models (like XGBoost) a continuous signal of how fast an item moves on average without needing high-cardinality one0hot encoding on raw SKU rating.</li>
    <li><mark><b>Item Price Elasticity Ratio</b></mark>: Signals pricing variation. A value < 1.0 indicates the item was sold at a discount relative to its normal average market price, where as > 1.0 indicates a premium charge.</li>
    <li><mark><b>Customer Unique Item Count</b></mark>: Measures account breadth over time. Customers with high item diversity buy across whole product portfolios (e.g., PLCs, Servos, HMIs) and have lower churn rates compared to single-item buyers.</li>

<p style='font-size: 1.4em;'><u><b style='color: red;'>Step 2.8.</b><b> Inventory Management Features</b></u></p>

<p style='font-size: 1.4em;'><u><b style='color: orange;'>Step 2.8.1.</b><b> 'Day Since Last Sale' Column</b></u></p>

```python
# Add Reference date (latest transaction in dataset).
ref_date = df['Sales Order Date'].max()

# Aggregate metrics at Item Level.
inventory_features = df.groupby(['Item Internal Id', 'Item Code']).agg(
    Last_Sale_Date=('Sales Order Date', 'max'),
    First_Sale_Date=('Sales Order Date', 'min'),
    Total_Unit_Sold=('Quantity', 'sum'),
    Total_Revenue=('Total Sales Amount', 'sum'),
    Order_Count=('So Internal Id', 'count')
).reset_index()

# Calculate Age & Inactivity Features.
inventory_features['Days_Since_Last_Sale'] = (ref_date - inventory_features['Last_Sale_Date']).dt.days
inventory_features['Years_Inactive'] = inventory_features['Days_Since_Last_Sale'] / 365.25
```

<p style='font-size: 1.4em;'><u><b style='color: orange;'>Step 2.8.2.</b><b> Inventory Velocity Status</b></u></p>

```python
def categorize_velocity(years):
    if years < 0.25:
        return 'Fast Moving (<3 Months)'
    elif years < 1.0:
        return 'Normal Velocity (3-12 Months)'
    elif years < 2.0:
        return 'Slow Moving (1-2 yrs)'
    elif years < 3.5:
        return 'High Aging Risk (2-3.5 yrs)'
    else:
        return 'Critical Risk! (>3.5 years)'

#
inventory_features['Inventory_Risk_Tier'] = inventory_features['Years_Inactive'].apply(categorize_velocity)

# Calculate total active days (clipping at minimum 1 to prevent division by zero)
inventory_features['Days_Active'] = (ref_date - inventory_features['First_Sale_Date']).dt.days.clip(lower=1)

# 1. Units sold per active day (Unit Velocity)
inventory_features['Sales_Frequency_Ratio_Units'] = inventory_features['Total_Unit_Sold'] / inventory_features['Days_Active']

# 2. Orders per active day (Order Velocity)
inventory_features['Sales_Frequency_Ratio_Orders'] = inventory_features['Order_Count'] / inventory_features['Days_Active']
inventory_features
```

<p style='font-size: 1.4em;'><u><b style='color: red;'>Step 2.9.</b><b> Dealing with Outliers</b></u></p>

```python
# Display top 5 largest values in the 'Total Sales Amount'.
df.nlargest(5, 'Total Sales Amount')
```

```python
# Apply percentile capping to numerical features.
cap_cols = ['Quantity', 'Total Sales Amount', 'Base Price']

for col in cap_cols:
    upper_limit = df[col].quantile(0.99)                # 99th percentile cutoff.
    lower_limit = df[col].quantile(0.001)               # Keep non-zero minimums.

    # Cap values beyond the 99th percentile.
    df[f'{col}_Capped'] = df[col].clip(lower=lower_limit, upper=upper_limit)

# Display row preserved.
print('Outlier capping complete. Row count preserved:', len(df))

df
```

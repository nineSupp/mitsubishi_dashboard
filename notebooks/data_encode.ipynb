## <b style='color: yellow;'>3.</b> <u>Data Encoding (For Categorical Data)</u>

            <li><b style='color: orange;'>Step 3.2.3.</b><b> One-Hot Encoding (For Low-Cardinality Nominal Categories)</b></li>

<p style='font-size: 1.4em;'><u><b style='color: red'>Step 3.1.</b> <b>Merge Two Data Together</u></b>

```python
# Select key item-level features from inventory file to prevent column duplications.
inv_cols = [
    'Item Internal Id',
    'Days_Since_Last_Sale',
    'Inventory_Risk_Tier',
    'Sales_Frequency_Ratio_Units',
    'Sales_Frequency_Ratio_Orders'
]
df_inv_subset = df_inv[inv_cols]


# Merge Inventory Context into Sales Data.
df = pd.merge(df_sales, df_inv_subset, on='Item Internal Id', how='left')

# Fill any missing inventory tiers for rare unmapped items.
df['Inventory_Risk_Tier'] = df['Inventory_Risk_Tier'].fillna('Unknown')
df
```

---

<p style='font-size: 1.4em;'><u><b style='color: red;'>Step 3.2.</b><b> Apply Categorical Encodings</b></u></p>

<p style='font-size: 1.4em;'><u><b style='color: orange;'>Step 3.2.1.</b><b> Ordinal Encoding (Preserve Clear Hierarchy)</b></u></p>

```python
# Mapping the 'Order Size Tier'.
order_tier_map = {
    'Small (<1k)': 1,
    'Medium (1k-10k)' : 2,
    'Large (10k-50k)': 3,
    'Enterprise (>50k)': 4
}

# Mapping the 'Inventory_Risk_Tier'.
inventory_risk_map = {
    "Fast Moving (<3 Months)": 1,
    "Normal Velocity (3-12 Months)": 2,
    "Slow Moving (1-2 yrs)": 3,
    "High Aging Risk (2-3.5 yrs)": 4,
    "Critical Risk! (>3.5 years)": 5,
    "Unknown": 0,
}

# Add column.
df['Order_Size_Tier_Encoded'] = (df['Order Size Tier'].map(order_tier_map).fillna(0))
df['Inventory_Risk_Tier_Encoded'] = (df['Inventory_Risk_Tier'].map(inventory_risk_map).fillna(0))
df
```

<p style='font-size: 1.4em;'><u><b style='color: orange;'>Step 3.2.2.</b><b> Frequency Encoding (For High-Cardinality IDs)</b></u></p>

```python
df['Customer_Order_Frequency'] = df.groupby('Customer Internal Id')['Customer Internal Id'].transform('count')
df['Item_Order_Frequency'] = df.groupby('Item Code')['Item Code'].transform('count')
```

<p style='font-size: 1.4em;'><u><b style='color: orange;'>Step 3.2.3.</b><b> One-Hot Encoding (For Low-Cardinality Nominal Categories)</b></u></p>

```python
ohe_col = [
    'Customer Type',
    'Production Status',
    'Sales Order Status Group',
    'Product Category 1',
    'Fa Sub Industries'
]

# Drop original nominal text columns and replace with dummy columns.
df_encode = pd.get_dummies(df, columns=ohe_col, drop_first=True, dtype=int)
df_encode
```

---

<p style='font-size: 1.4em;'><u><b style='color: red;'>Step 3.3.</b><b> Final Cleanup & Export to a New Excel File</b></u></p>

```python
# Drop raw string/date columns not usable directly by machine learning algorithms
columns_to_drop = [
    "Sales Order Status",
    "Company Name",
    "Product Category 2",
    "Year-Month",
    "Order Size Tier",
    "Inventory_Risk_Tier",
]

df_final = df_encode.drop(columns=[c for c in columns_to_drop if c in df_encode.columns])
df_final.to_excel('./excel/final_dataset_main.xlsx', index=False)
df_final.to_csv('./excel/final_dataset_main.csv', index=False)
```

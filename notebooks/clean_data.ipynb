<p style='font-size: 1.4em;'><u><b style='color: red'>Step 1.1.</b> <b>Load the 3 Excels Files</u></b><br>Load <mark><b>{Customer.xlsx</b>, <b>Item.xlsx</b>, <b>Sale Order History.xlsx}</b></mark> Files.</p>

```python
# 1) Load all 3 Excels Files.
files = ['Customer', 'Item', 'Sale Order History']
data = []

for file in files:
    data.append(pd.read_excel(f'./excel/{file}.xlsx'))
#============================================================================================================================================================================
# 2) Strip any Whitespace from column names across all df.

for df in data:
    df.columns = df.columns.str.strip()
#============================================================================================================================================================================
# display all 3 files, to make sure it's loaded correctly.
for dex, df in enumerate(data):
    print(f'{f'{files[dex]}.xlsx':=^180}')
    print(f'{df}')
    print(f'{'=' * 180}')
```

<p style='font-size: 1.4em;'><u><b style='color: red;'>1.2.</b> <b>Display The Attributes Name from each Dataset</b></u></p>

```python
# Find the name of the attributes for each dataset.
print(f'{f'Attributes/ Field Names of each dataset':=^180}')
for dex, df in enumerate(data):
    print(f'Current Files: {files[dex]}.xlsx')
    if dex == 2:
         print(f'{df.columns.tolist()}')
    else:
         print(f'{df.columns.tolist()}\n')
print(f'{'=' * 180}')
```

<p style='font-size: 1.4em;'><u><b style='color: red;'>1.3.</b> <b>Dealing with Duplicated Rows</b></u></p>

```python
# Find the number of duplicates from each Dataframe.
print(f'{f'Duplicate Rows':=^180}')
for dex, df in enumerate(data):
    print(f'Current Files: {files[dex]}.xlsx')
    if dex == 2:
        print(f'Have Numbers of Duplicates: {df.duplicated().sum()} duplicate rows.')
    else:
        print(f'Have Numbers of Duplicates: {df.duplicated().sum()} duplicate rows.\n')
print(f'{'=' * 180}')
```

<p style='font-size: 1.4em;'><b style='color: Green;'><u>NOTE!</u></b> Only the file <mark><b>Sale Order History.xlsx</b></mark> have duplicates, so we only dropping the duplicates for this dataset.</p>

```python
#NOTE:'Sales Order History.xlsx' is the only file that have duplicates. Thus, we remove the duplicates in this dataset.
# Use [-1] to quickly grab the last item in the list
data[-1] = data[-1].drop_duplicates()

if data[-1].duplicated().sum() == 0:
    print('The duplicates has been successfully removed')
else:
    print('The duplicate has not been successfully remove')
```

<p style='font-size: 1.4em;'><b style='color: Green;'><u>Recheck!</u></b> To see if there's still duplicates left.</p>

```python
print(f'{f'Duplicate Rows':=^180}')
for dex, df in enumerate(data):
    print(f'Current Files: {files[dex]}.xlsx')
    if dex == 2:
        print(f'Have Numbers of Duplicates: {df.duplicated().sum()} duplicate rows.')
    else:
        print(f'Have Numbers of Duplicates: {df.duplicated().sum()} duplicate rows.\n')
print(f'{'=' * 180}')
```

<p style='font-size: 1.4em;'><u><b style='color: red;'>Step 1.4.</b> <b>Dealing with Missing Values</b></u></p>

```python
# Find the number of missing values for each DataFrame.
print(f'{f'Missing Values':=^180}')
for dex, df in enumerate(data):
    print(f'{f'Current Files: {files[dex]}.xlsx':-^180}')
    print(f'{df.isna().sum()}')
    print(f'{'=' * 180}')
```

<p style='font-size: 1.4em;'><b style='color: Green;'><u>NOTE!</u></b> Only the data from <mark><b>'Item.xlsx'</b></mark> contains missing values.</p>

```python
# Merge or Join the data.
# Merge 1: Customer to Sale Order History.
sales_with_cus = pd.merge(
    data[-1],
    data[0][['Customer Internal ID', 'Company Name', 'Customer Type', 'FA Sub Industries']],
    on=['Customer Internal ID', 'Company Name'],
    how='left'
)

# Merge 2: Item to Merge 1.
final_df = pd.merge(
    sales_with_cus,
    data[1][['Item Internal ID', 'Item Code', 'Production Status', 'Product Category 1', 'Product Category 2', 'List price']],
    on=['Item Internal ID', 'Item Code'],
    how='left'
)

# Remove rows where both Base Price and List Price are 0 or missing.
zero_price_mask = (
    (final_df['Base Price'] == 0) & 
    ((final_df['List price'] == 0)) | (final_df['List price'].isna())
)
print(final_df.shape)
final_df = final_df[~zero_price_mask]
final_df
```

<p style='font-size: 1.4em;'><b style='color: green;'><u>Recheck!</u></b> Check for Missing or Duplicate Values</p>

```python
# Final Transformation.
print('=' * 180)
print(f'There are now: {final_df.duplicated().sum()} duplicates left.\n')
print(final_df.isna().sum())
print('=' * 180)
```

<p style='font-size: 1.4em;'><u><b style='color: red;'>Step 1.5.</b> <b>Change The Data Types to The Correct Attribute</b></u></p>

```python
# Display the info() of the DataFrame.
print('=' * 180)
final_df.info()
print('=' * 180)
```

<p style='font-size: 1.4em;'><b style='color: Green;'><u>NOTE!</u></b> The data type of all columns is correct!</p>

<p style='font-size: 1.4em;'><u><b style='color: red;'>Step 1.6.</b> <b> Clean the name of the column</b></u></p>

```python
# Display all the column names from the final dataframe.
print('=' * 180)
final_df.columns = final_df.columns.str.title()
for col in final_df.columns.tolist():
    print(col)
print('=' * 180)
```

<p style='font-size: 1.4em;'><u><b style='color: red;'>Step 1.7.</b> <b> Save to a New Excel File</b></u></p>

```python
# Save the cleaned data into the excel file.
final_df.to_excel('./excel/Cleaned_Sales.xlsx', index=False)
```
